```python
class AdvancedMotivationPatternExtractor:
    """ENHANCED realistic categorization system with domain expertise"""
    
    def __init__(self):
        # PROFESSIONAL TOURISM DOMAIN CATEGORIZATION - Enhanced with semantic analysis
        self.motivation_taxonomy = {
            'spatial_proximity': {
                'weight': 1.0,
                'patterns': [
                    r'\b(close|near|nearby|distance|walking|vicinity|adjacent|next to|meters?|km|minutes? walk|walking distance)\b',
                    r'\b(proximity|geographic|spatial|location|position|accessible|reachable)\b',
                    r'\b(short walk|few minutes|close by|around the corner|same area|same district)\b',
                    r'\b(within \d+\s*(m|meter|km|minute)|less than \d+\s*(m|meter|km|minute))\b'
                ],
                'semantic_keywords': ['distance', 'location', 'geography', 'space', 'proximity'],
                'label': 'Spatial Proximity',
                'description': 'Decisions based on physical distance and accessibility'
            },
            
            'temporal_optimization': {
                'weight': 0.95,
                'patterns': [
                    r'\b(time|hour|morning|afternoon|evening|schedule|timing|opening|closing|hours)\b',
                    r'\b(duration|period|availability|timetable|business hours|open|closed)\b',
                    r'\b(early|late|day|night|weekend|weekday|rush hour|peak time)\b',
                    r'\b(before closing|after opening|during lunch|free time|limited time)\b',
                    r'\b(\d+:\d+|am|pm|o\'clock|morning|afternoon|evening|night)\b'
                ],
                'semantic_keywords': ['time', 'schedule', 'temporal', 'timing', 'hours'],
                'label': 'Temporal Optimization',
                'description': 'Time-based decisions considering schedules and availability'
            },
            
            'cultural_thematic': {
                'weight': 0.9,
                'patterns': [
                    r'\b(historical|cultural|similar|pattern|preference|interest|theme|type|style)\b',
                    r'\b(consistent|continuity|same category|related|matching|alike|similar type)\b',
                    r'\b(art|history|culture|museum|religious|architectural|heritage|monument)\b',
                    r'\b(medieval|renaissance|baroque|gothic|ancient|modern|contemporary)\b',
                    r'\b(church|cathedral|palace|castle|fortress|villa|garden|archaeological)\b'
                ],
                'semantic_keywords': ['culture', 'history', 'art', 'heritage', 'theme'],
                'label': 'Cultural & Thematic',
                'description': 'Choices based on cultural interests and thematic continuity'
            },
            
            'destination_popularity': {
                'weight': 0.85,
                'patterns': [
                    r'\b(popular|famous|well-known|major attraction|tourist destination|must-see|iconic)\b',
                    r'\b(renowned|celebrated|notable|significant|important|landmark|highlight)\b',
                    r'\b(attraction|site|destination|main|primary|top|best|famous)\b',
                    r'\b(world heritage|unesco|world famous|internationally known)\b',
                    r'\b(guidebook|recommended|suggested|featured|mentioned)\b'
                ],
                'semantic_keywords': ['popularity', 'fame', 'attraction', 'tourism', 'reputation'],
                'label': 'Destination Popularity',
                'description': 'Selections based on popularity and tourist appeal'
            },
            
            'experiential_variety': {
                'weight': 0.8,
                'patterns': [
                    r'\b(different|variety|diverse|various|alternative|change|new experience)\b',
                    r'\b(explore|discovery|adventure|experience|journey|wandering)\b',
                    r'\b(tourist|visitor|sightseeing|tour|travel|trip|vacation|holiday)\b',
                    r'\b(explore more|see different|try something|variety of|diverse experience)\b',
                    r'\b(bucket list|must experience|unique|special|memorable)\b'
                ],
                'semantic_keywords': ['experience', 'variety', 'exploration', 'tourism', 'adventure'],
                'label': 'Experiential Variety',
                'description': 'Seeking diverse and enriching tourist experiences'
            },
            
            'quality_reputation': {
                'weight': 0.75,
                'patterns': [
                    r'\b(rating|review|recommend|quality|best|top|excellent|good|outstanding)\b',
                    r'\b(highly rated|well-reviewed|recommended|quality|superior|premium)\b',
                    r'\b(reputation|acclaimed|praised|award|recognized|certified|starred)\b',
                    r'\b(tripadvisor|michelin|lonely planet|frommer|fodor)\b',
                    r'\b(\d+\s*star|\d+/\d+|\d+\.\d+/\d+|five star|four star)\b'
                ],
                'semantic_keywords': ['quality', 'reputation', 'rating', 'recommendation', 'excellence'],
                'label': 'Quality & Reputation',
                'description': 'Choices influenced by ratings, reviews, and reputation'
            },
            
            'practical_convenience': {
                'weight': 0.7,
                'patterns': [
                    r'\b(convenient|easy|accessible|practical|simple|straightforward|comfortable)\b',
                    r'\b(efficient|quick|fast|direct|hassle-free|user-friendly|manageable)\b',
                    r'\b(public transport|metro|bus|tram|taxi|uber|parking|walkable)\b',
                    r'\b(facilities|restroom|restaurant|cafe|shop|services|amenities)\b',
                    r'\b(wheelchair|disabled|accessibility|barrier-free|family-friendly)\b'
                ],
                'semantic_keywords': ['convenience', 'accessibility', 'practical', 'transport', 'facilities'],
                'label': 'Practical Convenience',
                'description': 'Decisions based on practical considerations and convenience'
            },
            
            'social_behavioral': {
                'weight': 0.65,
                'patterns': [
                    r'\b(crowd|busy|quiet|peaceful|crowded|packed|popular with|less crowded)\b',
                    r'\b(group|family|friends|children|kids|elderly|couples|solo)\b',
                    r'\b(local|authentic|traditional|genuine|typical|characteristic)\b',
                    r'\b(atmosphere|ambiance|mood|feeling|vibe|energy|spirit)\b',
                    r'\b(social|community|people|locals|visitors|tourists)\b'
                ],
                'semantic_keywords': ['social', 'crowd', 'atmosphere', 'people', 'community'],
                'label': 'Social & Behavioral',
                'description': 'Influenced by social factors and crowd preferences'
            },
            
            'exclusion_filtering': {
                'weight': 0.6,
                'patterns': [
                    r'\b(already visited|not in|exclude|avoid|different|alternative|other than)\b',
                    r'\b(except|excluding|without|rather than|instead of|not interested)\b',
                    r'\b(skip|omit|bypass|alternative|different choice|not suitable)\b',
                    r'\b(been there|seen before|done that|repeat|duplicate|same as)\b',
                    r'\b(filter out|rule out|eliminate|dismiss|reject)\b'
                ],
                'semantic_keywords': ['exclusion', 'filter', 'avoid', 'different', 'alternative'],
                'label': 'Exclusion & Filtering',
                'description': 'Decisions to exclude or avoid certain options'
            },
            
            'economic_value': {
                'weight': 0.55,
                'patterns': [
                    r'\b(free|cost|price|expensive|cheap|affordable|budget|value)\b',
                    r'\b(money|euro|dollar|ticket|entrance fee|admission|discount)\b',
                    r'\b(worth|value for money|good deal|reasonable|economical)\b',
                    r'\b(student discount|senior discount|group discount|family ticket)\b',
                    r'\b(included|covered|additional cost|extra fee|no charge)\b'
                ],
                'semantic_keywords': ['cost', 'price', 'value', 'budget', 'economic'],
                'label': 'Economic Value',
                'description': 'Cost and value considerations in decision making'
            },
            
            'contextual_clustering': {
                'weight': 0.5,
                'patterns': [
                    r'\b(cluster|group|similar|pattern|category|type|classification)\b',
                    r'\b(same type|similar attractions|related sites|comparable|equivalent)\b',
                    r'\b(tourist cluster|attraction cluster|thematic group|category)\b',
                    r'\b(machine learning|algorithm|prediction|model|clustering)\b',
                    r'\b(data|analysis|pattern|correlation|similarity)\b'
                ],
                'semantic_keywords': ['cluster', 'group', 'similarity', 'pattern', 'category'],
                'label': 'Contextual Clustering',
                'description': 'Algorithmic grouping and pattern-based decisions'
            }
        }
        
        # ENHANCED FALLBACK CATEGORIES with domain expertise
        self.fallback_categories = {
            'location_geographical': {
                'patterns': [
                    r'\b(location|place|spot|area|zone|region|district|neighborhood)\b',
                    r'\b(position|situated|located|placed|coordinates|address)\b'
                ],
                'label': 'Location & Geography',
                'weight': 0.4
            },
            'user_preference': {
                'patterns': [
                    r'\b(user|tourist|traveler|person|individual|visitor|guest)\b',
                    r'\b(preference|choice|decision|selection|option|pick)\b'
                ],
                'label': 'User Preference',
                'weight': 0.35
            },
            'tourism_general': {
                'patterns': [
                    r'\b(tourism|tourist|travel|visit|trip|journey|vacation)\b',
                    r'\b(sightseeing|exploration|discovery|itinerary|tour)\b'
                ],
                'label': 'General Tourism',
                'weight': 0.3
            }
        }
        
        # SEMANTIC ENHANCEMENT: Context multipliers for realistic weighting
        self.context_multipliers = {
            'high_confidence_phrases': {
                'patterns': [
                    r'\b(specifically because|mainly due to|primarily because|the reason is)\b',
                    r'\b(based on|considering|given that|since|due to the fact)\b'
                ],
                'multiplier': 1.3
            },
            'uncertainty_phrases': {
                'patterns': [
                    r'\b(might|could|possibly|perhaps|maybe|potentially|likely)\b',
                    r'\b(seems|appears|looks like|probably|presumably)\b'
                ],
                'multiplier': 0.8
            },
            'negation_phrases': {
                'patterns': [
                    r'\b(not|isn\'t|wasn\'t|doesn\'t|won\'t|cannot|can\'t|never)\b',
                    r'\b(no|none|neither|nor|without|lacking)\b'
                ],
                'multiplier': 0.6
            }
        }
        
        # DOMAIN KNOWLEDGE: Tourism-specific context understanding
        self.tourism_contexts = {
            'verona_specific': {
                'patterns': [
                    r'\b(verona|romeo|juliet|arena|opera|amphitheater|adige|scaliger)\b',
                    r'\b(italian|italy|veneto|northern italy|venetian)\b'
                ],
                'multiplier': 1.2,
                'label_suffix': ' (Verona-specific)'
            },
            'italian_cultural': {
                'patterns': [
                    r'\b(renaissance|medieval|roman|italian|catholic|christian)\b',
                    r'\b(piazza|chiesa|palazzo|castello|ponte|torre|museo)\b'
                ],
                'multiplier': 1.1,
                'label_suffix': ' (Italian Cultural)'
            }
        }

    def analyze_semantic_context(self, text: str) -> Dict[str, float]:
        """Analyze semantic context for enhanced weighting"""
        text_lower = text.lower()
        context_scores = {'base': 1.0}
        
        # Apply confidence multipliers
        for context_type, config in self.context_multipliers.items():
            for pattern in config['patterns']:
                if re.search(pattern, text_lower):
                    context_scores[context_type] = config['multiplier']
                    break
        
        # Apply domain-specific multipliers
        for context_type, config in self.tourism_contexts.items():
            for pattern in config['patterns']:
                if re.search(pattern, text_lower):
                    context_scores[context_type] = config['multiplier']
                    break
        
        return context_scores

    def extract_motivations(self, reason_text: str) -> List[Dict]:
        """
        ENHANCED motivation extraction with semantic analysis and domain expertise
        
        Args:
            reason_text: Text to analyze
            
        Returns:
            List of dictionaries with enhanced motivation analysis
        """
        if pd.isna(reason_text) or str(reason_text).strip() == '':
            return [{'motivation': 'uncategorized', 'confidence': 0.0, 'matches': [], 'label': 'Uncategorized', 'reasoning': 'Empty or null text'}]
        
        text = str(reason_text).lower()
        extracted_motivations = []
        
        # SEMANTIC CONTEXT ANALYSIS
        context_scores = self.analyze_semantic_context(text)
        base_multiplier = np.prod(list(context_scores.values()))
        
        # PRIMARY TAXONOMY MATCHING with enhanced scoring
        for motivation, config in self.motivation_taxonomy.items():
            matches = []
            pattern_scores = []
            semantic_score = 0
            
            # Pattern matching with position weighting
            for i, pattern in enumerate(config['patterns']):
                pattern_matches = list(re.finditer(pattern, text, re.IGNORECASE))
                if pattern_matches:
                    # Weight early patterns higher (more specific)
                    pattern_weight = 1.0 - (i * 0.1)
                    pattern_weight = max(pattern_weight, 0.3)
                    
                    for match in pattern_matches:
                        matches.append(match.group())
                        # Position scoring: earlier matches get slight boost
                        position_score = 1.0 - (match.start() / len(text) * 0.1)
                        pattern_scores.append(pattern_weight * position_score)
            
            # Semantic keyword scoring
            for keyword in config.get('semantic_keywords', []):
                if keyword.lower() in text:
                    semantic_score += 0.15
            
            if matches or semantic_score > 0:
                # Calculate enhanced confidence
                pattern_confidence = sum(pattern_scores) * config['weight'] * 0.25
                total_confidence = (pattern_confidence + semantic_score) * base_multiplier
                
                # Normalize confidence to [0,1] range
                normalized_confidence = min(total_confidence, 1.0)
                
                # Determine domain context enhancement
                label = config['label']
                reasoning_details = []
                
                for context_type, context_config in self.tourism_contexts.items():
                    if context_type in context_scores:
                        label += context_config.get('label_suffix', '')
                        reasoning_details.append(f"Enhanced by {context_type}")
                        break
                
                extracted_motivations.append({
                    'motivation': motivation,
                    'confidence': round(normalized_confidence, 3),
                    'matches': list(set(matches)),
                    'label': label,
                    'semantic_score': round(semantic_score, 3),
                    'pattern_confidence': round(pattern_confidence, 3),
                    'context_multiplier': round(base_multiplier, 3),
                    'reasoning': f"{config['description']}. {'; '.join(reasoning_details)}" if reasoning_details else config['description']
                })
        
        # ENHANCED FALLBACK MATCHING with domain knowledge
        if not extracted_motivations:
            for category, config in self.fallback_categories.items():
                for pattern in config['patterns']:
                    pattern_matches = re.findall(pattern, text, re.IGNORECASE)
                    if pattern_matches:
                        confidence = len(pattern_matches) * config['weight'] * 0.3 * base_multiplier
                        confidence = min(confidence, 0.6)  # Cap fallback confidence
                        
                        extracted_motivations.append({
                            'motivation': category,
                            'confidence': round(confidence, 3),
                            'matches': list(set(pattern_matches)),
                            'label': config['label'],
                            'semantic_score': 0.0,
                            'pattern_confidence': round(confidence, 3),
                            'context_multiplier': round(base_multiplier, 3),
                            'reasoning': f"Fallback categorization based on {category} patterns"
                        })
                        break
        
        # ULTIMATE FALLBACK with enhanced reasoning
        if not extracted_motivations:
            extracted_motivations.append({
                'motivation': 'generic_reasoning',
                'confidence': 0.1 * base_multiplier,
                'matches': [],
                'label': 'Generic Reasoning',
                'semantic_score': 0.0,
                'pattern_confidence': 0.1,
                'context_multiplier': round(base_multiplier, 3),
                'reasoning': f"No specific patterns detected. Text length: {len(text)} chars. Context analysis applied."
            })
        
        # Sort by confidence and apply diversity penalty for multiple high-confidence matches
        extracted_motivations.sort(key=lambda x: x['confidence'], reverse=True)
        
        # DIVERSITY PENALTY: If multiple motivations have high confidence, reduce secondary ones
        if len(extracted_motivations) > 1:
            primary_confidence = extracted_motivations[0]['confidence']
            for i in range(1, len(extracted_motivations)):
                if extracted_motivations[i]['confidence'] > primary_confidence * 0.8:
                    # Reduce confidence of secondary high-confidence matches
                    extracted_motivations[i]['confidence'] *= 0.85
                    extracted_motivations[i]['confidence'] = round(extracted_motivations[i]['confidence'], 3)
        
        return extracted_motivations

    def batch_extract_enhanced(self, reason_texts: List[str]) -> List[List[Dict]]:
        """Enhanced batch processing with progress tracking and quality analysis"""
        results = []
        quality_metrics = {
            'high_confidence': 0,
            'medium_confidence': 0,
            'low_confidence': 0,
            'semantic_enhanced': 0,
            'domain_specific': 0
        }
        
        chunk_size = config.CHUNK_SIZE
        total_texts = len(reason_texts)
        
        logger.info(f"🧠 ENHANCED processing {total_texts:,} reasoning texts with domain expertise...")
        
        for i in tqdm(range(0, total_texts, chunk_size), desc="Enhanced extraction", ncols=100):
            chunk = reason_texts[i:i + chunk_size]
            chunk_results = []
            
            for text in chunk:
                motivations = self.extract_motivations(text)
                chunk_results.append(motivations)
                
                # Track quality metrics
                if motivations:
                    primary = motivations[0]
                    conf = primary['confidence']
                    
                    if conf >= 0.7:
                        quality_metrics['high_confidence'] += 1
                    elif conf >= 0.4:
                        quality_metrics['medium_confidence'] += 1
                    else:
                        quality_metrics['low_confidence'] += 1
                    
                    if primary.get('semantic_score', 0) > 0:
                        quality_metrics['semantic_enhanced'] += 1
                    
                    if 'specific' in primary['label'].lower():
                        quality_metrics['domain_specific'] += 1
            
            results.extend(chunk_results)
            
            # Memory management
            if (i // chunk_size + 1) % 10 == 0:
                gc.collect()
        
        # Log quality metrics
        logger.info(f"🎯 QUALITY ANALYSIS:")
        logger.info(f"   High confidence (≥0.7): {quality_metrics['high_confidence']:,} ({100*quality_metrics['high_confidence']/total_texts:.1f}%)")
        logger.info(f"   Medium confidence (0.4-0.7): {quality_metrics['medium_confidence']:,} ({100*quality_metrics['medium_confidence']/total_texts:.1f}%)")
        logger.info(f"   Low confidence (<0.4): {quality_metrics['low_confidence']:,} ({100*quality_metrics['low_confidence']/total_texts:.1f}%)")
        logger.info(f"   Semantic enhanced: {quality_metrics['semantic_enhanced']:,} ({100*quality_metrics['semantic_enhanced']/total_texts:.1f}%)")
        logger.info(f"   Domain specific: {quality_metrics['domain_specific']:,} ({100*quality_metrics['domain_specific']/total_texts:.1f}%)")
        
        return results


# ENHANCED PARALLEL PROCESSING FUNCTIONS - Updated to use AdvancedMotivationPatternExtractor

def process_file_parallel_enhanced(args: Tuple[Path, AdvancedMotivationPatternExtractor, CacheManager]) -> Dict:
    """
    Process a single file in parallel with ENHANCED categorization - with caching support
    
    Args:
        args: Tuple of (csv_file, enhanced_extractor, cache_manager)
        
    Returns:
        Dictionary with file processing results including enhanced metrics
    """
    csv_file, extractor, cache_manager = args
    
    # Check cache first
    if config.ENABLE_CACHING and cache_manager.is_cached(csv_file):
        cached_result = cache_manager.load_cache(csv_file)
        if cached_result:
            return {
                'file_name': csv_file.name,
                'status': 'cached',
                'motivation_counts': cached_result['motivation_counts'],
                'confidence_sums': cached_result['confidence_sums'],
                'total_records': cached_result['total_records'],
                'quality_metrics': cached_result.get('quality_metrics', {})
            }
    
    # Process file if not cached
    try:
        motivation_counts = defaultdict(int)
        confidence_sums = defaultdict(float)
        quality_metrics = {
            'high_confidence': 0,
            'semantic_enhanced': 0,
            'domain_specific': 0,
            'avg_confidence': 0.0
        }
        total_records = 0
        confidence_total = 0.0
        
        # Read file in chunks
        for chunk in pd.read_csv(csv_file, chunksize=config.CHUNK_SIZE, 
                               encoding='utf-8', low_memory=False, 
                               on_bad_lines='skip', dtype={'reason': 'object'}):
            
            # Filter valid records
            if 'status' in chunk.columns:
                chunk = chunk[chunk['status'] == 'success']
            
            if 'reason' not in chunk.columns:
                continue
            
            # Filter valid reasons
            valid_reasons = chunk[
                chunk['reason'].notna() & 
                (chunk['reason'].astype(str).str.len() >= config.MIN_REASON_LENGTH)
            ]['reason'].tolist()
            
            if not valid_reasons:
                continue
            
            # Process motivations with enhanced extraction
            for reason_text in valid_reasons:
                motivations = extractor.extract_motivations(reason_text)
                if motivations:
                    primary_motivation = motivations[0]
                    motivation_key = primary_motivation['label']
                    confidence = primary_motivation['confidence']
                    
                    motivation_counts[motivation_key] += 1
                    confidence_sums[motivation_key] += confidence
                    confidence_total += confidence
                    total_records += 1
                    
                    # Enhanced quality tracking
                    if confidence >= 0.7:
                        quality_metrics['high_confidence'] += 1
                    
                    if primary_motivation.get('semantic_score', 0) > 0:
                        quality_metrics['semantic_enhanced'] += 1
                    
                    if 'specific' in motivation_key.lower() or 'verona' in motivation_key.lower():
                        quality_metrics['domain_specific'] += 1
        
        # Calculate average confidence
        if total_records > 0:
            quality_metrics['avg_confidence'] = round(confidence_total / total_records, 3)
        
        # Prepare result
        result = {
            'file_name': csv_file.name,
            'status': 'processed',
            'motivation_counts': dict(motivation_counts),
            'confidence_sums': dict(confidence_sums),
            'total_records': total_records,
            'quality_metrics': quality_metrics
        }
        
        # Save to cache
        if config.ENABLE_CACHING:
            cache_data = {
                'motivation_counts': result['motivation_counts'],
                'confidence_sums': result['confidence_sums'],
                'total_records': result['total_records'],
                'quality_metrics': result['quality_metrics']
            }
            cache_manager.save_cache(csv_file, cache_data)
        
        return result
        
    except Exception as e:
        return {
            'file_name': csv_file.name,
            'status': 'error',
            'error': str(e),
            'motivation_counts': {},
            'confidence_sums': {},
            'total_records': 0,
            'quality_metrics': {}
        }


def parallel_analysis_with_enhanced_categorization(results_dirs: List[str] = None) -> pd.DataFrame:
    """
    ULTRA-FAST PARALLEL ANALYSIS with ENHANCED REALISTIC CATEGORIZATION
    
    Args:
        results_dirs: Directories to scan
        
    Returns:
        Complete analysis results DataFrame
    """
    if results_dirs is None:
        results_dirs = ["../results/middle", "../results/penultimate"]
    
    logger.info("🧠 ENHANCED CATEGORIZATION ANALYSIS STARTING")
    logger.info(f"Workers: {config.N_WORKERS}")
    logger.info("🎯 Features: 11 specialized categories + semantic analysis + domain expertise")
    
    # Initialize enhanced managers
    cache_manager = CacheManager(config.CACHE_DIR, config.CACHE_VERSION + "_enhanced")
    checkpoint_manager = CheckpointManager(config.CHECKPOINT_FILE + "_enhanced")
    
    # Show cache stats
    cache_stats = cache_manager.get_cache_stats()
    logger.info(f"📊 Cache status: {cache_stats['cached_files']} files cached ({cache_stats['cache_size_mb']} MB)")
    
    # Discover all files
    all_csv_files = []
    for results_dir in results_dirs:
        results_path = Path(results_dir)
        if results_path.exists():
            csv_files = list(results_path.rglob('*pred*.csv'))
            csv_files = [f for f in csv_files if 'checkpoint' not in f.name and f.stat().st_size > 0]
            all_csv_files.extend(csv_files)
    
    if not all_csv_files:
        raise FileNotFoundError(f"No CSV files found in {results_dirs}")
    
    logger.info(f"📁 Total files discovered: {len(all_csv_files)}")
    
    # Check for existing checkpoint
    checkpoint_data = None
    if config.CHECKPOINT_ENABLED:
        checkpoint_data = checkpoint_manager.load_checkpoint()
    
    # Initialize or resume progress
    if checkpoint_data:
        processed_files = set(checkpoint_data['processed_files'])
        aggregated_results = checkpoint_data['aggregated_results']
        remaining_files = [f for f in all_csv_files if f.name not in processed_files]
        
        logger.info(f"🔄 RESUMING: {len(processed_files)} files already done, {len(remaining_files)} remaining")
    else:
        processed_files = set()
        remaining_files = all_csv_files
        aggregated_results = {
            'motivation_counts': defaultdict(int),
            'confidence_sums': defaultdict(float),
            'total_records': 0,
            'quality_metrics': {
                'high_confidence': 0,
                'semantic_enhanced': 0,
                'domain_specific': 0,
                'total_confidence': 0.0
            }
        }
        
        logger.info(f"🆕 STARTING FRESH: {len(remaining_files)} files to process")
    
    # Quick check for cached files
    cached_count = sum(1 for f in remaining_files if cache_manager.is_cached(f))
    logger.info(f"⚡ Speed boost: {cached_count}/{len(remaining_files)} files already cached")
    
    # Initialize ENHANCED extractor
    extractor = AdvancedMotivationPatternExtractor()
    
    # Process files in parallel
    processed_count = len(processed_files)
    failed_count = 0
    
    if remaining_files:
        logger.info(f"🧠 ENHANCED PARALLEL PROCESSING: {len(remaining_files)} files with {config.N_WORKERS} workers")
        logger.info("🎯 Categories: Spatial Proximity, Temporal Optimization, Cultural Thematic, Destination Popularity...")
        
        # Prepare arguments for parallel processing
        process_args = [(csv_file, extractor, cache_manager) for csv_file in remaining_files]
        
        # Execute parallel processing
        with ThreadPoolExecutor(max_workers=config.N_WORKERS) as executor:
            # Submit all tasks
            future_to_file = {
                executor.submit(process_file_parallel_enhanced, args): args[0] 
                for args in process_args
            }
            
            # Process results as they complete
            with tqdm(total=len(remaining_files), desc="Enhanced parallel processing", ncols=100) as pbar:
                for future in as_completed(future_to_file):
                    csv_file = future_to_file[future]
                    
                    try:
                        result = future.result()
                        
                        # Update aggregated results
                        for motivation, count in result['motivation_counts'].items():
                            aggregated_results['motivation_counts'][motivation] += count
                        
                        for motivation, conf_sum in result['confidence_sums'].items():
                            aggregated_results['confidence_sums'][motivation] += conf_sum
                        
                        aggregated_results['total_records'] += result['total_records']
                        
                        # Update quality metrics
                        file_quality = result.get('quality_metrics', {})
                        aggregated_results['quality_metrics']['high_confidence'] += file_quality.get('high_confidence', 0)
                        aggregated_results['quality_metrics']['semantic_enhanced'] += file_quality.get('semantic_enhanced', 0)
                        aggregated_results['quality_metrics']['domain_specific'] += file_quality.get('domain_specific', 0)
                        aggregated_results['quality_metrics']['total_confidence'] += file_quality.get('avg_confidence', 0) * result['total_records']
                        
                        processed_files.add(result['file_name'])
                        processed_count += 1
                        
                        if result['status'] == 'error':
                            failed_count += 1
                            logger.debug(f"❌ {result['file_name']}: {result['error']}")
                        elif result['status'] == 'cached':
                            logger.debug(f"⚡ {result['file_name']}: loaded from cache")
                        
                        # Save checkpoint periodically
                        if config.CHECKPOINT_ENABLED and processed_count % config.CHECKPOINT_INTERVAL == 0:
                            progress_info = {
                                'processed_count': processed_count,
                                'total_files': len(all_csv_files),
                                'failed_count': failed_count
                            }
                            checkpoint_manager.save_checkpoint(
                                list(processed_files), 
                                aggregated_results, 
                                progress_info
                            )
                        
                        pbar.update(1)
                        
                        # Update progress info
                        if processed_count % 25 == 0:
                            cache_hit_rate = cached_count / len(remaining_files) * 100 if remaining_files else 0
                            logger.info(f"📈 Progress: {processed_count}/{len(all_csv_files)} files ({processed_count/len(all_csv_files)*100:.1f}%) | Cache hit: {cache_hit_rate:.1f}%")
                    
                    except Exception as e:
                        logger.error(f"❌ Error processing {csv_file.name}: {e}")
                        failed_count += 1
                        pbar.update(1)
    
    # Calculate enhanced results
    total_records = aggregated_results['total_records']
    motivation_data = []
    
    for motivation, count in aggregated_results['motivation_counts'].items():
        percentage = (count / total_records * 100) if total_records > 0 else 0
        avg_confidence = (aggregated_results['confidence_sums'][motivation] / count) if count > 0 else 0
        
        motivation_data.append({
            'motivation_category': motivation,
            'count': count,
            'percentage': round(percentage, 2),
            'avg_confidence': round(avg_confidence, 3),
            'source_file': 'enhanced_analysis'
        })
    
    # Create results DataFrame
    results_df = pd.DataFrame(motivation_data)
    results_df = results_df.sort_values('count', ascending=False)
    
    # Calculate enhanced quality metrics
    quality_metrics = aggregated_results['quality_metrics']
    avg_confidence = quality_metrics['total_confidence'] / total_records if total_records > 0 else 0
    
    # Clear checkpoint on successful completion
    if config.CHECKPOINT_ENABLED and failed_count == 0:
        checkpoint_manager.clear_checkpoint()
    
    # Final cache stats
    final_cache_stats = cache_manager.get_cache_stats()
    
    logger.info("=== ENHANCED CATEGORIZATION ANALYSIS COMPLETED ===")
    logger.info(f"🧠 TOTAL RECORDS: {total_records:,}")
    logger.info(f"📊 FILES PROCESSED: {processed_count}/{len(all_csv_files)}")
    logger.info(f"📊 SUCCESS RATE: {(processed_count-failed_count)/processed_count*100:.1f}%")
    logger.info(f"⚡ CACHE PERFORMANCE: {final_cache_stats['cached_files']} files cached ({final_cache_stats['cache_size_mb']} MB)")
    
    # Enhanced quality reporting
    logger.info(f"🎯 QUALITY METRICS:")
    logger.info(f"   Average Confidence: {avg_confidence:.3f}")
    logger.info(f"   High Confidence (≥0.7): {quality_metrics['high_confidence']:,} ({100*quality_metrics['high_confidence']/total_records:.1f}%)")
    logger.info(f"   Semantic Enhanced: {quality_metrics['semantic_enhanced']:,} ({100*quality_metrics['semantic_enhanced']/total_records:.1f}%)")
    logger.info(f"   Domain Specific: {quality_metrics['domain_specific']:,} ({100*quality_metrics['domain_specific']/total_records:.1f}%)")
    
    # Show top motivations
    logger.info(f"🏆 TOP 5 ENHANCED MOTIVATIONS:")
    for i, row in results_df.head().iterrows():
        logger.info(f"   {i+1}. {row['motivation_category']:30s}: {row['count']:8,} ({row['percentage']:5.1f}%) [conf: {row['avg_confidence']:.3f}]")
    
    return results_df


print("✅ ENHANCED MOTIVATION PATTERN EXTRACTOR READY!")
print("🧠 Advanced semantic analysis with tourism domain expertise")
print("🎯 11 specialized motivation categories with realistic weighting")
print("📍 Verona-specific and Italian cultural context awareness")
print("⚡ Quality metrics and confidence analysis")
print("🔍 Enhanced pattern matching with position and semantic scoring")
```

```python
def preprocess_and_validate_reasoning_data(df: pd.DataFrame) -> Tuple[pd.DataFrame, Dict]:
    """
    Professional preprocessing with comprehensive quality assessment
    
    Args:
        df: Raw DataFrame from data loading
        
    Returns:
        Tuple of (cleaned_df, quality_metrics)
    """
    logger.info("Starting data preprocessing and quality assessment...")
    
    initial_count = len(df)
    quality_metrics = {
        'initial_records': initial_count,
        'processing_steps': [],
        'data_quality_flags': []
    }
    
    # Step 1: Basic data validation
    logger.info("Step 1: Basic data validation")
    if 'reason' not in df.columns:
        # Try to find reason-like columns
        reason_cols = [col for col in df.columns if 'reason' in col.lower()]
        if reason_cols:
            logger.info(f"Using column '{reason_cols[0]}' as reason column")
            df = df.rename(columns={reason_cols[0]: 'reason'})
        else:
            raise ValueError("No 'reason' column found in the dataset")
    
    # Step 2: Filter by status if available
    logger.info("Step 2: Status filtering")
    if 'status' in df.columns:
        success_count = (df['status'] == 'success').sum()
        df_clean = df[df['status'] == 'success'].copy()
        logger.info(f"Filtered by status: {len(df_clean):,}/{initial_count:,} success records ({100*len(df_clean)/initial_count:.1f}%)")
        quality_metrics['processing_steps'].append(f"Status filtering: {success_count:,} success records")
    else:
        df_clean = df.copy()
        quality_metrics['data_quality_flags'].append("No status column available")
    
    # Step 3: Reason text validation
    logger.info("Step 3: Reason text validation")
    
    # Remove null/empty reasons
    before_null = len(df_clean)
    df_clean = df_clean[df_clean['reason'].notna()].copy()
    df_clean = df_clean[df_clean['reason'].astype(str).str.strip() != ''].copy()
    after_null = len(df_clean)
    
    # Length filtering
    df_clean['reason_length'] = df_clean['reason'].astype(str).str.len()
    before_length = len(df_clean)
    df_clean = df_clean[df_clean['reason_length'] >= config.MIN_REASON_LENGTH].copy()
    after_length = len(df_clean)
    
    logger.info(f"Reason validation:")
    logger.info(f"  - Null/empty removal: {after_null:,}/{before_null:,} ({100*after_null/before_null:.1f}%)")
    logger.info(f"  - Length filtering (≥{config.MIN_REASON_LENGTH} chars): {after_length:,}/{before_length:,} ({100*after_length/before_length:.1f}%)")
    
    quality_metrics['processing_steps'].extend([
        f"Null/empty removal: {after_null:,} records retained",
        f"Length filtering (≥{config.MIN_REASON_LENGTH} chars): {after_length:,} records retained"
    ])
    
    # Step 4: Language and content quality assessment
    logger.info("Step 4: Content quality assessment")
    
    # Basic content quality checks
    df_clean['reason_word_count'] = df_clean['reason'].astype(str).str.split().str.len()
    df_clean['reason_has_punctuation'] = df_clean['reason'].astype(str).str.contains(r'[.!?]')
    df_clean['reason_has_keywords'] = df_clean['reason'].astype(str).str.lower().str.contains(r'\\b(tourist|visit|location|distance|attraction|popular|time|preference)\\b')
    
    # Quality scoring
    quality_scores = []
    for _, row in df_clean.iterrows():
        score = 0
        reason_text = str(row['reason']).lower()
        
        # Length score (0-3 points)
        if row['reason_length'] >= 50:
            score += 3
        elif row['reason_length'] >= 30:
            score += 2
        elif row['reason_length'] >= 15:
            score += 1
        
        # Word count score (0-2 points)
        if row['reason_word_count'] >= 10:
            score += 2
        elif row['reason_word_count'] >= 5:
            score += 1
        
        # Content quality (0-3 points)
        if row['reason_has_punctuation']:
            score += 1
        if row['reason_has_keywords']:
            score += 2
        
        # Language quality (simple heuristics)
        if re.search(r'\\b(the|and|for|with|from|because|due|based|near|close|popular)\\b', reason_text):
            score += 1
        
        quality_scores.append(score)
    
    df_clean['quality_score'] = quality_scores
    
    # Quality distribution
    quality_dist = df_clean['quality_score'].describe()
    logger.info(f"Quality score distribution: Mean={quality_dist['mean']:.1f}, Std={quality_dist['std']:.1f}")
    
    # Step 5: Final dataset summary
    logger.info("=== PREPROCESSING SUMMARY ===")
    final_count = len(df_clean)
    retention_rate = (final_count / initial_count) * 100
    
    logger.info(f"Records retained: {final_count:,}/{initial_count:,} ({retention_rate:.1f}%)")
    logger.info(f"Average reason length: {df_clean['reason_length'].mean():.1f} characters")
    logger.info(f"Average word count: {df_clean['reason_word_count'].mean():.1f} words")
    logger.info(f"Quality score range: {df_clean['quality_score'].min()}-{df_clean['quality_score'].max()}")
    
    # Update quality metrics
    quality_metrics.update({
        'final_records': final_count,
        'retention_rate': retention_rate,
        'avg_reason_length': float(df_clean['reason_length'].mean()),
        'avg_word_count': float(df_clean['reason_word_count'].mean()),
        'quality_score_stats': {
            'mean': float(quality_dist['mean']),
            'std': float(quality_dist['std']),
            'min': int(df_clean['quality_score'].min()),
            'max': int(df_clean['quality_score'].max())
        }
    })
    
    if 'strategy' in df_clean.columns:
        strategy_counts = df_clean['strategy'].value_counts()
        logger.info("Final strategy distribution:")
        for strategy, count in strategy_counts.items():
            logger.info(f"  {strategy}: {count:,} records ({100*count/final_count:.1f}%)")
        quality_metrics['strategy_distribution'] = strategy_counts.to_dict()
    
    print(f"✅ PREPROCESSING COMPLETED")
    print(f"📊 Final dataset: {final_count:,} high-quality reasoning texts")
    print(f"📊 Retention rate: {retention_rate:.1f}%")
    print(f"📊 Average quality score: {quality_dist['mean']:.1f}/10")
    
    return df_clean, quality_metrics

# Execute preprocessing
try:
    df_processed, quality_metrics = preprocess_and_validate_reasoning_data(df_raw)
    
    # Store total records count for final reporting
    TOTAL_RECORDS = len(df_processed)
    
    print(f"\\n📈 TOTAL RECORDS FOR ANALYSIS: {TOTAL_RECORDS:,}")
    
except Exception as e:
    logger.error(f"Error in preprocessing: {str(e)}")
    print(f"❌ PREPROCESSING ERROR: {str(e)}")
    raise
```

```python
class MotivationPatternExtractor:
    """MEMORY OPTIMIZED motivation pattern extraction with advanced NLP techniques"""
    
    def __init__(self):
        # Professional motivation taxonomy with weighted patterns (NO NLTK required)
        self.motivation_taxonomy = {
            'geographic_proximity': {
                'weight': 1.0,
                'patterns': [
                    r'\b(close|near|nearby|distance|walking|vicinity|adjacent|next to|km away|meters|minutes walk)\b',
                    r'\b(proximity|geographic|spatial|location|position)\b',
                    r'\b(reachable|accessible|convenient location)\b'
                ],
                'label': 'Geographic Proximity'
            },
            'popularity_attraction': {
                'weight': 0.9,
                'patterns': [
                    r'\b(popular|famous|well-known|major attraction|tourist destination|must-see|iconic)\b',
                    r'\b(renowned|celebrated|notable|significant|important|landmark)\b',
                    r'\b(attraction|site|destination|highlight|main)\b'
                ],
                'label': 'Popularity & Attraction'
            },
            'thematic_continuity': {
                'weight': 0.8,
                'patterns': [
                    r'\b(historical|cultural|similar|pattern|preference|interest|theme|type)\b',
                    r'\b(consistent|continuity|same category|related|matching)\b',
                    r'\b(art|history|culture|museum|religious|architectural)\b'
                ],
                'label': 'Thematic Continuity'
            },
            'temporal_factors': {
                'weight': 0.7,
                'patterns': [
                    r'\b(time|hour|morning|afternoon|evening|schedule|timing|opening|closing)\b',
                    r'\b(duration|period|availability|hours|timetable)\b',
                    r'\b(early|late|day|night|weekend|weekday)\b'
                ],
                'label': 'Temporal Factors'
            },
            'tourist_experience': {
                'weight': 0.85,
                'patterns': [
                    r'\b(tourist|visitor|experience|tour|sightseeing|explore|visit|journey)\b',
                    r'\b(traveler|trip|vacation|holiday|itinerary|route)\b',
                    r'\b(exploration|discovery|adventure|experience)\b'
                ],
                'label': 'Tourist Experience'
            },
            'exclusion_logic': {
                'weight': 0.6,
                'patterns': [
                    r'\b(already visited|not in|exclude|avoid|different|alternative|other)\b',
                    r'\b(except|excluding|without|rather than|instead of)\b',
                    r'\b(skip|omit|bypass|alternative|different choice)\b'
                ],
                'label': 'Exclusion Logic'
            },
            'quality_rating': {
                'weight': 0.75,
                'patterns': [
                    r'\b(rating|review|recommend|quality|best|top|excellent|good)\b',
                    r'\b(highly rated|well-reviewed|recommended|quality|superior)\b',
                    r'\b(reputation|acclaimed|praised|award|recognized)\b'
                ],
                'label': 'Quality & Rating'
            },
            'convenience_practical': {
                'weight': 0.65,
                'patterns': [
                    r'\b(convenient|easy|accessible|practical|simple|straightforward)\b',
                    r'\b(efficient|quick|fast|direct|comfortable)\b',
                    r'\b(hassle-free|user-friendly|manageable)\b'
                ],
                'label': 'Convenience & Practical'
            }
        }
        
        # Fallback categories for unmatched patterns
        self.fallback_categories = {
            'location_based': {
                'patterns': [r'\b(location|place|spot|area|zone|region)\b'],
                'label': 'Location Based'
            },
            'user_focused': {
                'patterns': [r'\b(user|tourist|traveler|person|individual)\b'],
                'label': 'User Focused'
            }
        }
    
    def extract_motivations(self, reason_text: str) -> List[Dict]:
        """
        Extract motivation patterns with confidence scores (NO NLTK required)
        
        Args:
            reason_text: Text to analyze
            
        Returns:
            List of dictionaries with motivation, confidence, and matches
        """
        if pd.isna(reason_text) or str(reason_text).strip() == '':
            return [{'motivation': 'uncategorized', 'confidence': 0.0, 'matches': [], 'label': 'Uncategorized'}]
        
        text = str(reason_text).lower()
        extracted_motivations = []
        
        # Primary taxonomy matching
        for motivation, config in self.motivation_taxonomy.items():
            matches = []
            total_confidence = 0.0
            
            for pattern in config['patterns']:
                pattern_matches = re.findall(pattern, text, re.IGNORECASE)
                if pattern_matches:
                    matches.extend(pattern_matches)
                    # Confidence based on match frequency and pattern weight
                    confidence = len(pattern_matches) * config['weight'] * 0.3
                    total_confidence += confidence
            
            if matches:
                # Normalize confidence to [0,1] range
                normalized_confidence = min(total_confidence, 1.0)
                extracted_motivations.append({
                    'motivation': motivation,
                    'confidence': normalized_confidence,
                    'matches': list(set(matches)),  # Remove duplicates
                    'label': config['label']
                })
        
        # Fallback matching if no primary matches
        if not extracted_motivations:
            for category, config in self.fallback_categories.items():
                for pattern in config['patterns']:
                    if re.search(pattern, text, re.IGNORECASE):
                        extracted_motivations.append({
                            'motivation': category,
                            'confidence': 0.3,
                            'matches': re.findall(pattern, text, re.IGNORECASE),
                            'label': config['label']
                        })
                        break
        
        # Final fallback
        if not extracted_motivations:
            extracted_motivations.append({
                'motivation': 'generic_reasoning',
                'confidence': 0.1,
                'matches': [],
                'label': 'Generic Reasoning'
            })
        
        # Sort by confidence
        extracted_motivations.sort(key=lambda x: x['confidence'], reverse=True)
        
        return extracted_motivations
    
    def batch_extract_chunked(self, reason_texts: List[str]) -> List[List[Dict]]:
        """MEMORY OPTIMIZED batch processing with chunking"""
        results = []
        chunk_size = config.CHUNK_SIZE
        total_texts = len(reason_texts)
        
        logger.info(f"Processing {total_texts:,} reasoning texts in chunks of {chunk_size:,}...")
        
        for i in tqdm(range(0, total_texts, chunk_size), desc="Processing chunks", ncols=100):
            chunk = reason_texts[i:i + chunk_size]
            chunk_results = []
            
            for text in chunk:
                motivations = self.extract_motivations(text)
                chunk_results.append(motivations)
            
            results.extend(chunk_results)
            
            # MEMORY MANAGEMENT: Force garbage collection every chunk
            if (i // chunk_size + 1) % 10 == 0:
                gc.collect()
        
        return results

# Initialize and execute MEMORY OPTIMIZED extraction
logger.info("Initializing MEMORY OPTIMIZED motivation pattern extractor...")
extractor = MotivationPatternExtractor()

# Extract motivations with memory optimization
logger.info("Starting memory-optimized motivation extraction...")
motivation_results = extractor.batch_extract_chunked(df_processed['reason'].tolist())

# Add results to dataframe
df_processed['motivation_extractions'] = motivation_results

# Create primary motivation column (highest confidence)
df_processed['primary_motivation'] = [
    results[0]['motivation'] if results else 'uncategorized' 
    for results in motivation_results
]

df_processed['primary_motivation_label'] = [
    results[0]['label'] if results else 'Uncategorized' 
    for results in motivation_results
]

df_processed['primary_confidence'] = [
    results[0]['confidence'] if results else 0.0 
    for results in motivation_results
]

# Memory cleanup
del motivation_results
gc.collect()

logger.info("✅ MEMORY OPTIMIZED motivation extraction completed successfully!")
print(f"✅ MEMORY OPTIMIZED MOTIVATION EXTRACTION COMPLETED")
print(f"📊 Processed {len(df_processed):,} reasoning texts")
print(f"📊 Average confidence: {df_processed['primary_confidence'].mean():.3f}")
print(f"📊 High confidence predictions (>0.5): {(df_processed['primary_confidence'] > 0.5).sum():,} ({100*(df_processed['primary_confidence'] > 0.5).sum()/len(df_processed):.1f}%)")
print(f"📊 Memory usage: {df_processed.memory_usage(deep=True).sum() / 1024**2:.1f} MB")
```

## 4. Statistical Analysis and Results Aggregation

```python
def generate_comprehensive_analysis(df: pd.DataFrame) -> Tuple[pd.DataFrame, Dict]:
    """
    Generate comprehensive statistical analysis of motivation patterns
    
    Args:
        df: Processed DataFrame with motivation extractions
        
    Returns:
        Tuple of (analysis_results_df, statistical_summary)
    """
    logger.info("Generating comprehensive statistical analysis...")
    
    # Primary motivation distribution
    motivation_counts = df['primary_motivation'].value_counts()
    motivation_labels = df.groupby('primary_motivation')['primary_motivation_label'].first()
    
    total_records = len(df)
    
    # Create detailed analysis results
    analysis_data = []
    cumulative_percentage = 0
    
    for motivation, count in motivation_counts.items():
        percentage = (count / total_records) * 100
        cumulative_percentage += percentage
        
        # Calculate confidence statistics for this motivation
        motivation_df = df[df['primary_motivation'] == motivation]
        confidence_stats = motivation_df['primary_confidence'].describe()
        
        analysis_data.append({
            'Motivation_Code': motivation,
            'Motivation_Label': motivation_labels[motivation],
            'Count': int(count),
            'Percentage': round(percentage, 2),
            'Cumulative_Percentage': round(cumulative_percentage, 2),
            'Avg_Confidence': round(confidence_stats['mean'], 3),
            'Min_Confidence': round(confidence_stats['min'], 3),
            'Max_Confidence': round(confidence_stats['max'], 3),
            'Std_Confidence': round(confidence_stats['std'], 3)
        })
    
    analysis_df = pd.DataFrame(analysis_data)
    
    # Generate comprehensive statistical summary
    statistical_summary = {
        'dataset_overview': {
            'total_records': total_records,
            'unique_motivations': len(motivation_counts),
            'analysis_timestamp': datetime.now().isoformat(),
            'most_common_motivation': {
                'code': motivation_counts.index[0],
                'label': motivation_labels[motivation_counts.index[0]],
                'count': int(motivation_counts.iloc[0]),
                'percentage': round((motivation_counts.iloc[0] / total_records) * 100, 2)
            },
            'least_common_motivation': {
                'code': motivation_counts.index[-1],
                'label': motivation_labels[motivation_counts.index[-1]],
                'count': int(motivation_counts.iloc[-1]),
                'percentage': round((motivation_counts.iloc[-1] / total_records) * 100, 2)
            }
        },
        'confidence_metrics': {
            'overall_avg_confidence': round(df['primary_confidence'].mean(), 3),
            'overall_std_confidence': round(df['primary_confidence'].std(), 3),
            'high_confidence_count': int((df['primary_confidence'] > 0.5).sum()),
            'high_confidence_percentage': round(((df['primary_confidence'] > 0.5).sum() / total_records) * 100, 2),
            'confidence_distribution': {
                'very_high_0.8+': int((df['primary_confidence'] >= 0.8).sum()),
                'high_0.5-0.8': int(((df['primary_confidence'] >= 0.5) & (df['primary_confidence'] < 0.8)).sum()),
                'medium_0.3-0.5': int(((df['primary_confidence'] >= 0.3) & (df['primary_confidence'] < 0.5)).sum()),
                'low_0.0-0.3': int((df['primary_confidence'] < 0.3).sum())
            }
        }
    }
    
    # Add strategy-based analysis if available
    if 'strategy' in df.columns:
        strategy_analysis = {}
        for strategy in df['strategy'].unique():
            strategy_df = df[df['strategy'] == strategy]
            strategy_motivation_counts = strategy_df['primary_motivation'].value_counts()
            
            strategy_analysis[strategy] = {
                'record_count': len(strategy_df),
                'top_motivation': {
                    'code': strategy_motivation_counts.index[0],
                    'label': motivation_labels[strategy_motivation_counts.index[0]],
                    'count': int(strategy_motivation_counts.iloc[0]),
                    'percentage': round((strategy_motivation_counts.iloc[0] / len(strategy_df)) * 100, 2)
                },
                'avg_confidence': round(strategy_df['primary_confidence'].mean(), 3)
            }
        
        statistical_summary['strategy_analysis'] = strategy_analysis
    
    # Quality assessment
    if 'quality_score' in df.columns:
        quality_stats = df['quality_score'].describe()
        statistical_summary['quality_assessment'] = {
            'avg_quality_score': round(quality_stats['mean'], 2),
            'std_quality_score': round(quality_stats['std'], 2),
            'high_quality_count': int((df['quality_score'] >= 7).sum()),
            'high_quality_percentage': round(((df['quality_score'] >= 7).sum() / total_records) * 100, 2)
        }
    
    logger.info("Statistical analysis completed successfully")
    logger.info(f"Generated analysis for {len(analysis_df)} motivation categories")
    logger.info(f"Top motivation: {statistical_summary['dataset_overview']['most_common_motivation']['label']} ({statistical_summary['dataset_overview']['most_common_motivation']['percentage']:.1f}%)")
    
    return analysis_df, statistical_summary

# Execute comprehensive analysis
analysis_results_df, stats_summary = generate_comprehensive_analysis(df_processed)

# Display key insights
print("✅ COMPREHENSIVE STATISTICAL ANALYSIS COMPLETED")
print(f"📊 Total Records Analyzed: {stats_summary['dataset_overview']['total_records']:,}")
print(f"📊 Unique Motivation Categories: {stats_summary['dataset_overview']['unique_motivations']}")
print(f"📊 Most Common Motivation: {stats_summary['dataset_overview']['most_common_motivation']['label']} ({stats_summary['dataset_overview']['most_common_motivation']['percentage']:.1f}%)")
print(f"📊 Average Confidence Score: {stats_summary['confidence_metrics']['overall_avg_confidence']:.3f}")
print(f"📊 High Confidence Predictions: {stats_summary['confidence_metrics']['high_confidence_percentage']:.1f}%")

print("\\n🏆 TOP 5 MOTIVATION CATEGORIES:")
for i, row in analysis_results_df.head().iterrows():
    print(f"   {i+1}. {row['Motivation_Label']:25s}: {row['Count']:6,} ({row['Percentage']:5.1f}%) [conf: {row['Avg_Confidence']:.3f}]")
```
