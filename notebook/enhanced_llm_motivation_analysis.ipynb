```python
# CELL 2: Advanced Configuration and Performance Settings

class AnalysisConfig:
    """High-performance analysis configuration for 25M+ records"""
    
    def __init__(self):
        # Performance Settings
        self.MAX_WORKERS = min(mp.cpu_count(), 16)  # Optimal for A100 setup
        self.CHUNK_SIZE = 50000  # Records per chunk
        self.BATCH_SIZE = 10000  # Processing batch size
        self.MEMORY_LIMIT_GB = 30  # Memory usage limit
        
        # Analysis Settings
        self.MIN_CONFIDENCE = 0.3  # Minimum confidence threshold
        self.MAX_CATEGORIES = 15  # Maximum motivation categories
        self.SEMANTIC_THRESHOLD = 0.7  # Semantic similarity threshold
        
        # File Paths
        self.RESULTS_BASE_DIR = Path('/leonardo_work/IscrC_LLM-Mob/LLM-Mob-As-Mobility-Interpreter/results')
        self.CACHE_DIR = Path('./cache/motivation_analysis')
        self.OUTPUT_DIR = Path('./output/enhanced_motivation_analysis')
        
        # Ensure directories exist
        self.CACHE_DIR.mkdir(parents=True, exist_ok=True)
        self.OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        
        # Visualization Settings
        self.FIGURE_SIZE = (24, 16)
        self.DPI = 300
        self.COLOR_PALETTE = 'husl'
        
        logger.info(f"Configuration initialized: {self.MAX_WORKERS} workers, {self.CHUNK_SIZE} chunk size")

class PerformanceMonitor:
    """Real-time performance monitoring for large-scale processing"""
    
    def __init__(self):
        self.start_time = None
        self.processed_records = 0
        self.memory_usage = []
        self.processing_rates = []
        
    def start(self):
        self.start_time = time.time()
        self.processed_records = 0
        logger.info("Performance monitoring started")
    
    def update(self, records_processed: int):
        self.processed_records += records_processed
        current_memory = psutil.virtual_memory().percent
        self.memory_usage.append(current_memory)
        
        if self.start_time:
            elapsed = time.time() - self.start_time
            rate = self.processed_records / elapsed if elapsed > 0 else 0
            self.processing_rates.append(rate)
            
            if self.processed_records % 100000 == 0:  # Log every 100k records
                logger.info(f"Processed {self.processed_records:,} records at {rate:.0f} records/sec, Memory: {current_memory:.1f}%")
    
    def get_stats(self) -> Dict:
        elapsed = time.time() - self.start_time if self.start_time else 0
        return {
            'total_records': self.processed_records,
            'elapsed_time': elapsed,
            'avg_rate': self.processed_records / elapsed if elapsed > 0 else 0,
            'max_memory': max(self.memory_usage) if self.memory_usage else 0,
            'peak_rate': max(self.processing_rates) if self.processing_rates else 0
        }

# Initialize configuration
config = AnalysisConfig()
monitor = PerformanceMonitor()

print(f"⚙️ High-performance configuration loaded")
print(f"🔧 Workers: {config.MAX_WORKERS}, Chunk size: {config.CHUNK_SIZE:,}")
print(f"💾 Memory limit: {config.MEMORY_LIMIT_GB}GB")
print(f"📁 Cache directory: {config.CACHE_DIR}")
```

```python
# CELL 3: Advanced Motivation Pattern Extractor with Deep Analysis

class ComprehensiveMotivationAnalyzer:
    """Advanced LLM motivation analysis with semantic understanding"""
    
    def __init__(self):
        # Enhanced Motivation Categories (15 specialized categories)
        self.motivation_patterns = {
            'spatial_proximity': {
                'patterns': [
                    r'(close|near|nearby|proxim|adjacent|walking distance)',
                    r'(short distance|easy to reach|next to|around the corner)',
                    r'(same area|vicinity|neighborhood|region)'
                ],
                'semantic_keywords': ['proximity', 'distance', 'location', 'nearby', 'accessibility'],
                'weight': 1.0
            },
            'cultural_heritage': {
                'patterns': [
                    r'(cultural|historic|heritage|monument|tradition)',
                    r'(museum|gallery|church|cathedral|basilica)',
                    r'(art|architecture|ancient|medieval|renaissance)'
                ],
                'semantic_keywords': ['culture', 'history', 'art', 'heritage', 'monument'],
                'weight': 1.2
            },
            'thematic_continuity': {
                'patterns': [
                    r'(similar|same theme|continuation|related)',
                    r'(complement|matches interest|follows pattern)',
                    r'(consistent with|aligns with|thematic)'
                ],
                'semantic_keywords': ['theme', 'continuity', 'pattern', 'consistency', 'similarity'],
                'weight': 1.1
            },
            'temporal_logic': {
                'patterns': [
                    r'(time|schedule|opening hours|timing)',
                    r'(morning|afternoon|evening|lunch|dinner)',
                    r'(available now|open|accessible at this time)'
                ],
                'semantic_keywords': ['time', 'schedule', 'timing', 'availability', 'hours'],
                'weight': 1.0
            },
            'user_preference': {
                'patterns': [
                    r'(prefer|interest|like|enjoy|favorite)',
                    r'(personal|user|visitor|tourist preference)',
                    r'(based on history|past visits|previous choices)'
                ],
                'semantic_keywords': ['preference', 'interest', 'personal', 'choice', 'history'],
                'weight': 1.3
            },
            'social_popular': {
                'patterns': [
                    r'(popular|famous|well-known|renowned|iconic)',
                    r'(must-see|recommended|top-rated|highly rated)',
                    r'(tourist attraction|main site|landmark)'
                ],
                'semantic_keywords': ['popular', 'famous', 'recommended', 'attraction', 'landmark'],
                'weight': 0.9
            },
            'practical_convenience': {
                'patterns': [
                    r'(convenient|practical|easy|accessible|available)',
                    r'(transportation|parking|facilities|services)',
                    r'(no reservation|immediate access|walk-in)'
                ],
                'semantic_keywords': ['convenient', 'practical', 'accessible', 'facilities', 'service'],
                'weight': 0.8
            },
            'experiential_diversity': {
                'patterns': [
                    r'(different|variety|diverse|new experience)',
                    r'(change of pace|contrast|alternative)',
                    r'(explore|discover|try something new)'
                ],
                'semantic_keywords': ['different', 'variety', 'diverse', 'experience', 'explore'],
                'weight': 1.1
            },
            'seasonal_temporal': {
                'patterns': [
                    r'(season|weather|climate|outdoor|indoor)',
                    r'(spring|summer|autumn|winter|seasonal)',
                    r'(temperature|rain|sun|shade)'
                ],
                'semantic_keywords': ['seasonal', 'weather', 'climate', 'outdoor', 'indoor'],
                'weight': 0.7
            },
            'quality_reputation': {
                'patterns': [
                    r'(quality|excellent|outstanding|exceptional)',
                    r'(reputation|renowned|prestigious|acclaimed)',
                    r'(award|recognition|rating|review)'
                ],
                'semantic_keywords': ['quality', 'reputation', 'excellent', 'prestigious', 'acclaimed'],
                'weight': 1.0
            },
            'economic_value': {
                'patterns': [
                    r'(free|cost|price|budget|affordable)',
                    r'(discount|offer|deal|promotion)',
                    r'(included|card benefit|no extra cost)'
                ],
                'semantic_keywords': ['free', 'cost', 'budget', 'value', 'affordable'],
                'weight': 0.9
            },
            'educational_learning': {
                'patterns': [
                    r'(learn|educational|informative|knowledge)',
                    r'(understand|discover|explore|study)',
                    r'(guide|tour|explanation|interpretation)'
                ],
                'semantic_keywords': ['educational', 'learning', 'knowledge', 'informative', 'guide'],
                'weight': 1.0
            },
            'verona_specific': {
                'patterns': [
                    r'(verona|veneto|romeo|juliet|shakespeare)',
                    r'(arena|amphitheater|colosseum|roman)',
                    r'(adige|scaliger|medieval|venetian)'
                ],
                'semantic_keywords': ['verona', 'romeo', 'juliet', 'arena', 'roman'],
                'weight': 1.5  # Higher weight for location-specific reasoning
            },
            'social_group': {
                'patterns': [
                    r'(family|children|kids|group|couple)',
                    r'(friends|social|together|shared experience)',
                    r'(suitable for|appropriate for|good for)'
                ],
                'semantic_keywords': ['family', 'group', 'social', 'together', 'suitable'],
                'weight': 1.0
            },
            'serendipity_random': {
                'patterns': [
                    r'(random|chance|spontaneous|unexpected)',
                    r'(serendipity|surprise|unplanned|impulse)',
                    r'(why not|just because|no specific reason)'
                ],
                'semantic_keywords': ['random', 'spontaneous', 'surprise', 'unplanned', 'impulse'],
                'weight': 0.5
            }
        }
        
        # Compile regex patterns for performance
        self.compiled_patterns = {}
        for category, data in self.motivation_patterns.items():
            self.compiled_patterns[category] = [
                re.compile(pattern, re.IGNORECASE | re.MULTILINE)
                for pattern in data['patterns']
            ]
        
        # Initialize TF-IDF for semantic analysis
        self.tfidf = TfidfVectorizer(
            max_features=1000,
            stop_words='english',
            ngram_range=(1, 2),
            min_df=2
        )
        
        logger.info(f"Motivation analyzer initialized with {len(self.motivation_patterns)} categories")
    
    def analyze_single_reason(self, reason_text: str) -> Dict[str, Any]:
        """Deep analysis of a single reasoning text"""
        if not reason_text or pd.isna(reason_text):
            return {'category': 'unknown', 'confidence': 0.0, 'matches': []}
        
        reason_text = str(reason_text).lower().strip()
        category_scores = {}
        detailed_matches = {}
        
        for category, patterns in self.compiled_patterns.items():
            category_data = self.motivation_patterns[category]
            matches = []
            total_score = 0.0
            
            # Pattern matching with position weighting
            for i, pattern in enumerate(patterns):
                for match in pattern.finditer(reason_text):
                    match_text = match.group()
                    # Weight matches by position (earlier = more important)
                    position_weight = 1.0 - (match.start() / len(reason_text)) * 0.3
                    score = category_data['weight'] * position_weight
                    total_score += score
                    matches.append({
                        'text': match_text,
                        'position': match.start(),
                        'score': score
                    })
            
            # Semantic keyword matching
            semantic_score = 0.0
            for keyword in category_data['semantic_keywords']:
                if keyword in reason_text:
                    semantic_score += 0.5 * category_data['weight']
            
            total_score += semantic_score
            
            if total_score > 0:
                category_scores[category] = total_score
                detailed_matches[category] = matches
        
        if not category_scores:
            return {'category': 'unknown', 'confidence': 0.0, 'matches': []}
        
        # Determine best category with confidence
        best_category = max(category_scores, key=category_scores.get)
        max_score = category_scores[best_category]
        total_score = sum(category_scores.values())
        confidence = min(max_score / total_score, 1.0) if total_score > 0 else 0.0
        
        return {
            'category': best_category,
            'confidence': confidence,
            'raw_score': max_score,
            'matches': detailed_matches.get(best_category, []),
            'all_scores': category_scores,
            'text_length': len(reason_text),
            'match_density': len(detailed_matches) / len(reason_text) if reason_text else 0
        }
    
    def batch_analyze(self, reason_texts: List[str]) -> List[Dict[str, Any]]:
        """High-performance batch analysis"""
        results = []
        
        with ThreadPoolExecutor(max_workers=config.MAX_WORKERS) as executor:
            futures = {
                executor.submit(self.analyze_single_reason, text): idx 
                for idx, text in enumerate(reason_texts)
            }
            
            for future in tqdm(as_completed(futures), total=len(futures), desc="Analyzing motivations"):
                result = future.result()
                results.append(result)
                
                if len(results) % 10000 == 0:
                    monitor.update(10000)
        
        return results

print("🧠 Comprehensive Motivation Analyzer initialized")
print(f"📊 Analysis categories: {len(ComprehensiveMotivationAnalyzer().motivation_patterns)}")
print("⚡ Ready for high-performance batch processing")
```

```python
# CELL 4: High-Performance Data Loading and Preprocessing

class DataProcessor:
    """Optimized data loading and preprocessing for large datasets"""
    
    def __init__(self, config: AnalysisConfig):
        self.config = config
        self.analyzer = ComprehensiveMotivationAnalyzer()
    
    def discover_result_files(self) -> List[Path]:
        """Auto-discover all result files in the directory tree"""
        result_files = []
        
        # Search patterns for different model results
        search_patterns = [
            '**/with_geom_time/*_pred_*.csv',
            '**/with_geom/*_pred_*.csv', 
            '**/base_version/*_pred_*.csv'
        ]
        
        for pattern in search_patterns:
            files = list(self.config.RESULTS_BASE_DIR.rglob(pattern))
            result_files.extend(files)
            logger.info(f"Found {len(files)} files matching pattern: {pattern}")
        
        logger.info(f"Total discovered files: {len(result_files)}")
        return sorted(list(set(result_files)))  # Remove duplicates and sort
    
    def load_file_optimized(self, file_path: Path) -> pd.DataFrame:
        """Memory-optimized file loading with chunking"""
        try:
            # Detect file characteristics
            file_size_mb = file_path.stat().st_size / (1024 * 1024)
            
            if file_size_mb > 500:  # Large file
                logger.info(f"Loading large file ({file_size_mb:.1f}MB) with chunking: {file_path.name}")
                chunks = []
                chunksize = self.config.CHUNK_SIZE
                
                for chunk in pd.read_csv(file_path, chunksize=chunksize):
                    # Filter only rows with reasoning data
                    if 'reason' in chunk.columns:
                        chunk_filtered = chunk[chunk['reason'].notna()]
                        if not chunk_filtered.empty:
                            chunks.append(chunk_filtered)
                    
                    # Memory management
                    current_memory = psutil.virtual_memory().percent
                    if current_memory > 85:  # High memory usage
                        gc.collect()
                        logger.warning(f"High memory usage ({current_memory:.1f}%), triggering garbage collection")
                
                df = pd.concat(chunks, ignore_index=True) if chunks else pd.DataFrame()
            else:
                df = pd.read_csv(file_path)
                if 'reason' in df.columns:
                    df = df[df['reason'].notna()]
            
            logger.info(f"Loaded {len(df):,} records from {file_path.name}")
            return df
            
        except Exception as e:
            logger.error(f"Error loading file {file_path}: {e}")
            return pd.DataFrame()
    
    def extract_metadata(self, file_path: Path) -> Dict[str, str]:
        """Extract model and strategy metadata from file path"""
        parts = file_path.parts
        metadata = {
            'model': 'unknown',
            'strategy': 'unknown',
            'dataset': 'unknown',
            'file_path': str(file_path)
        }
        
        # Extract model name
        model_patterns = ['qwen2.5_7b', 'qwen2.5_14b', 'llama3.1_8b', 'mixtral_8x7b', 'mistral_7b', 'deepseek-coder_33b']
        for part in parts:
            for model in model_patterns:
                if model in part.lower():
                    metadata['model'] = model
                    break
        
        # Extract strategy
        if 'with_geom_time' in str(file_path):
            metadata['strategy'] = 'geospatial_temporal'
        elif 'with_geom' in str(file_path):
            metadata['strategy'] = 'geospatial_only'
        elif 'base_version' in str(file_path):
            metadata['strategy'] = 'base_version'
        
        # Extract dataset year
        import re
        year_match = re.search(r'dati_(\d{4})', file_path.name)
        if year_match:
            metadata['dataset'] = f"verona_{year_match.group(1)}"
        
        return metadata
    
    def preprocess_reasons(self, df: pd.DataFrame) -> Tuple[pd.DataFrame, Dict]:
        """Advanced preprocessing of reasoning texts"""
        stats = {
            'total_records': len(df),
            'valid_reasons': 0,
            'avg_reason_length': 0,
            'empty_reasons': 0
        }
        
        if 'reason' not in df.columns:
            logger.warning("No 'reason' column found in dataframe")
            return df, stats
        
        # Clean and validate reasons
        df['reason_clean'] = df['reason'].astype(str).str.strip()
        df['reason_clean'] = df['reason_clean'].replace('nan', '')
        df['reason_length'] = df['reason_clean'].str.len()
        
        # Filter valid reasons (min 10 characters)
        valid_mask = (df['reason_length'] >= 10) & (df['reason_clean'] != '')
        df_valid = df[valid_mask].copy()
        
        stats.update({
            'valid_reasons': len(df_valid),
            'avg_reason_length': df_valid['reason_length'].mean() if len(df_valid) > 0 else 0,
            'empty_reasons': len(df) - len(df_valid)
        })
        
        logger.info(f"Preprocessed {stats['valid_reasons']:,} valid reasons from {stats['total_records']:,} records")
        return df_valid, stats

# Initialize data processor
data_processor = DataProcessor(config)
print("🔄 High-performance data processor initialized")
print(f"📁 Scanning results directory: {config.RESULTS_BASE_DIR}")
```

```python
# CELL 5: Ultra-High Performance Parallel Analysis Engine

def process_file_chunk(args: Tuple[Path, int, int]) -> Dict[str, Any]:
    """Process a single file chunk in parallel"""
    file_path, start_idx, end_idx = args
    
    try:
        # Load data chunk
        df = pd.read_csv(file_path, skiprows=range(1, start_idx), nrows=end_idx-start_idx)
        
        if 'reason' not in df.columns or df.empty:
            return {'file': str(file_path), 'results': [], 'error': 'No reason column or empty data'}
        
        # Initialize analyzer for this process
        analyzer = ComprehensiveMotivationAnalyzer()
        
        # Process reasoning texts
        reasons = df['reason'].dropna().astype(str).tolist()
        results = analyzer.batch_analyze(reasons)
        
        return {
            'file': str(file_path),
            'chunk': f"{start_idx}-{end_idx}",
            'results': results,
            'processed_count': len(results)
        }
    except Exception as e:
        return {'file': str(file_path), 'chunk': f"{start_idx}-{end_idx}", 'error': str(e), 'results': []}

def parallel_motivation_analysis(max_files: int = None) -> Dict[str, Any]:
    """Ultra-high performance parallel analysis of all result files"""
    
    # Start performance monitoring
    monitor.start()
    start_time = time.time()
    
    # Discover all result files
    result_files = data_processor.discover_result_files()
    if max_files:
        result_files = result_files[:max_files]
    
    logger.info(f"Starting parallel analysis of {len(result_files)} files")
    
    # Aggregate results storage
    all_motivation_results = defaultdict(list)
    file_metadata = []
    processing_stats = {
        'total_files': len(result_files),
        'successful_files': 0,
        'total_records_processed': 0,
        'errors': []
    }
    
    # Process files in parallel with chunking
    with ProcessPoolExecutor(max_workers=config.MAX_WORKERS) as executor:
        # Prepare file chunks for processing
        chunk_jobs = []
        
        for file_path in result_files:
            try:
                # Estimate file size for chunking
                file_size = file_path.stat().st_size
                estimated_rows = min(file_size // 100, 1000000)  # Rough estimation
                
                if estimated_rows > config.CHUNK_SIZE:
                    # Split large files into chunks
                    num_chunks = (estimated_rows // config.CHUNK_SIZE) + 1
                    for i in range(num_chunks):
                        start_idx = i * config.CHUNK_SIZE + 1  # +1 to skip header
                        end_idx = min((i + 1) * config.CHUNK_SIZE, estimated_rows)
                        chunk_jobs.append((file_path, start_idx, end_idx))
                else:
                    chunk_jobs.append((file_path, 1, estimated_rows))
                    
            except Exception as e:
                logger.error(f"Error preparing file {file_path}: {e}")
                processing_stats['errors'].append(f"File preparation error: {file_path} - {e}")
        
        logger.info(f"Created {len(chunk_jobs)} processing chunks")
        
        # Submit all jobs
        future_to_job = {
            executor.submit(process_file_chunk, job): job 
            for job in chunk_jobs
        }
        
        # Process results as they complete
        progress_bar = tqdm(total=len(future_to_job), desc="Processing file chunks")
        
        for future in as_completed(future_to_job):
            try:
                result = future.result(timeout=300)  # 5-minute timeout per chunk
                
                if 'error' in result:
                    logger.error(f"Chunk processing error: {result['error']}")
                    processing_stats['errors'].append(result['error'])
                else:
                    # Aggregate results
                    file_path = Path(result['file'])
                    metadata = data_processor.extract_metadata(file_path)
                    
                    for motivation_result in result['results']:
                        category = motivation_result['category']
                        confidence = motivation_result['confidence']
                        
                        all_motivation_results[category].append({
                            'confidence': confidence,
                            'model': metadata['model'],
                            'strategy': metadata['strategy'],
                            'dataset': metadata['dataset']
                        })
                    
                    processing_stats['total_records_processed'] += result['processed_count']
                    monitor.update(result['processed_count'])
                
            except Exception as e:
                job = future_to_job[future]
                logger.error(f"Future processing error for {job}: {e}")
                processing_stats['errors'].append(f"Future error: {job} - {e}")
            
            progress_bar.update(1)
            
            # Memory management
            if psutil.virtual_memory().percent > 90:
                gc.collect()
                logger.warning("High memory usage, triggering garbage collection")
        
        progress_bar.close()
    
    # Calculate final statistics
    total_time = time.time() - start_time
    performance_stats = monitor.get_stats()
    
    # Create comprehensive results summary
    motivation_summary = []
    total_records = sum(len(records) for records in all_motivation_results.values())
    
    for category, records in all_motivation_results.items():
        if records:  # Only include categories with data
            avg_confidence = np.mean([r['confidence'] for r in records])
            count = len(records)
            percentage = (count / total_records) * 100 if total_records > 0 else 0
            
            motivation_summary.append({
                'motivation_category': category,
                'count': count,
                'percentage': percentage,
                'avg_confidence': avg_confidence,
                'records': records  # Keep detailed records for further analysis
            })
    
    # Sort by count (descending)
    motivation_summary.sort(key=lambda x: x['count'], reverse=True)
    
    results = {
        'motivation_summary': motivation_summary,
        'processing_stats': processing_stats,
        'performance_stats': performance_stats,
        'total_processing_time': total_time,
        'total_records': total_records
    }
    
    # Log final results
    logger.info(f"Analysis completed in {total_time:.2f} seconds")
    logger.info(f"Processed {total_records:,} total records")
    logger.info(f"Average processing rate: {performance_stats['avg_rate']:.0f} records/second")
    logger.info(f"Identified {len(motivation_summary)} motivation categories")
    
    return results

print("🚀 Ultra-high performance parallel analysis engine ready")
print(f"⚡ Configuration: {config.MAX_WORKERS} workers, {config.CHUNK_SIZE:,} chunk size")
print("🎯 Ready to process 25M+ records efficiently")
```
